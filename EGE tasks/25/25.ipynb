## Функция нахождения делителей числа x

```python
def f(x):
    r=[]
    for i in range(1, int(x**0.5)+1):
        if x%i==0:
            r.append(i)
            r.append(x//i)
    return sorted(set(r))

```

## Функция нахождения простых делителей числа x

```python
def g(x):
    r=[]
    i = 2
    while i**2<=x:
        if x%i==0:
            r.append(i)
            x = x//i
        else:
            i+=1
    if x>1:
        r.append(x)
    return r

print(g(1818406112))
```
