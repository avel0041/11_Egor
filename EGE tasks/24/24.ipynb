```python
f=open('24 (1).txt').readline()
m=0
l = 0
k9 = 0

for r in range(1, len(f)):
    if f[r-1:r+1]=='K9':
        k9+=1
    while k9>150:
        if f[l-1:l+1]=='K9':
            k9-=1
        l+=1
    m = max(m, r-l+2)
print(m)
```
